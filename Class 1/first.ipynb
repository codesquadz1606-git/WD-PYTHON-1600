print() it is a builtin function which is used to print the statement , operation or a variable.

```python
print("Hello")
print(10+10)
```

end="" is a function , which helps us to convert block statement into inline statement

```python
print("Hello",end=" ")
print("How are you?",end=" ")
print("I am Fine",end=" ")
print("What about you?",end=" ")
```

\n : it means break the line from their.
\t : returns us the tab space.

```python
print("Hello \nHow are You?")
print("Hello\tHow are you?")
```

f"statement {operation}" : format string

```python
# print("The sum of 10 & 10 is",10+10,"\nMultiplication of 10 * 10 is",10*10)
print(f"The sum of 10 & 10 is {10+10} \nMultiplcation of 10 * 10 is {10*10}")
```
