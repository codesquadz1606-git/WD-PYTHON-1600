What is set?
It stores unique elements in unordered list.It is mutable as well as immutable.

```python
# new_set={10,20,30,40,10,20,59,10}
# print(new_set)
# print(type(new_set))

set1=set()

# set1.add(20)
set1.update([10,20,30,40,50,60,70])
set1.pop()
set1.pop()
set1.pop()

set1.remove(60)
# set1.remove(60) # it delivers the error if element is not found
set1.discard(60) # it do not give any error if element o=is not found.

print(set1)
# print(type(set1))
```

```python
# How sets are immutable?
set2={10,20,30,40,50,60}
fs=frozenset(set2) # i frozed my set
fs.add(100) # now set has turned immutable.
print(fs)
```

```python
# Mathematical
s1={10,20,30,40,50}
s2={40,50,60,70,80}
# union
print(s1.union(s2))

# intersection
print(s1.intersection(s2))

# difference
print(s1.difference(s2))
print(s2.difference(s1))

# symmetric-diffrence
print(s1.symmetric_difference(s2))
```
