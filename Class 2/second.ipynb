What is Data Type?
It identify which type of value is stored in a variable.
Types: 
int(Non-decimal number , +ve or -ve) , 
float(Decimal number , +ve or -ve) , 
string(Any value inside "") , 
boolean(True or False) , 
none(no type for now.)

```python
a=10
b=98.5
c="Ayush123"
d=True
e=None

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

Type Casting : It means conversion of one data type to another.

```python
a="10"
print(type(a))
a=int(a)
print(type(a))
a=float(a)
print(type(a))
a=str(a)
print(type(a))
```

What is operators?
Operator means operation b/w two or more operands.

A+B=AB
here , A & B is operand , + is operator & AB is operation.

Types :
i. Arthemetic : +,-,*,/,//,%,**
ii. Relational : <,>,<=,>=,==,!=
iii. Assignment : =,+=,-=,*=,/=,%=,**=
iv. Logical : and , or , not
v. Membership : in , not in
vi. Identify : is , not is

```python
# Arthemetic
a=10
b=3

print(a+b)
print(a-b)
print(a*b)
print(a/b) # Quotient
print(a//b) # floor division :removes the decimals.
print(a%b) # Remainder
print(a**b)
```

```python
# Relational Operator
a=10
b=15

print(a<b) # T
print(a>b) # F
print(a<=b) # T
print(a>=b) # F
print(a==b) # F
print(a!=b) # T
```

```python
# Assignment Operator : Assigning + Arthemetic
a=10 # 10 has been assigned to a
a+=10 # a = a+10
a-=10
a*=3
a/=5
a%=4
a**=3
print(a)
```

```python
# Logical Operator : and , or , not

a=10
b=15
c=9

# and : It means that all the condition must be true , then only final result will be true
# login form : where username & password both must be correct.
print(a<b and b<c)

# or : Here , if any single condition is true , then final result is true.
# fingerprint : phone , attendance
print(a<b or b<c)

# not : Reversal true--> false , false--> true
print(not(a<b or b<c))
```
