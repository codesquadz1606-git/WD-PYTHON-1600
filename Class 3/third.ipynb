COnditional Statement : Here , we print any statement based on the condition.

if & else
if : it is the starting statement
else : it is the ending statement

syntax
if(condition):
    statement
else:
    statement

```python
# age>=19 : Eligible , age<=18 : Not Eligible.

age=int(input("Enter age"))
if(age>=19):
    print("Eligible")
elif(age>=12 and age<=18):
    print(f"Eligible in {19-age} years")
else:
    print("Not Eligible")
```

marks : input
marks : 90 greater : A+
marks : >70 & <90 : B
marks : >55 & <70 : C
marks : >45 & <55 : D
marks : >33 & <45 : E
marks : <33  : F

```python
marks=int(input("Enter the marks"))

if(marks>=90):
    print("Grade : A+")
elif(marks>=70 and marks<90):
    print("Grade : B")
elif(marks>=55 and marks<70):
    print("Grade : C")
elif(marks>=45 and marks<55):
    print("Grade : D")
elif(marks>=33 and marks<45):
    print("Grade : E")
else:
    print("Grade : F")
```
