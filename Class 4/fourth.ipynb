Nested If Else

find the greatest b/w three numbers : a, b, c

```python
a=int(input("Enter a : "))
b=int(input("Enter b : "))
c=int(input("Enter c : "))

if(a>b):
    if(a>c):
        print("A is greatest")
    else:
        print("C is greatest")
elif(b>c):
    print("B is greatest")
else:
    print("C is greatest")
```

Upi ka max transaction allowed : 100000
Daily limit : 200000



```python
tax_amount=int(input("Enter the amount to spend"))
today_spent=int(input("Enter the today's spent amount"))

MAX_ALLOW_TRANSACTION=100000
UPI_DAILY_LIMIT=200000

if(tax_amount<=MAX_ALLOW_TRANSACTION and (today_spent+tax_amount)<=UPI_DAILY_LIMIT):
    print("Transcation Successfull")
else:
    print("Transaction Rejected : As you have crossed your daily limit.")
```

5pm to 6am or raining : 1.5x increase ho jata h

```python
distance=float(input("Enter the distance"))
hour=int(input("Enter the time"))
is_raining=False

fare=distance*12

if(hour>=6 and hour<=17) or not(is_raining):
    print("Fare:",fare)
else:
    revised_fare=fare*1.5
    print("Fare:",revised_fare)
```
