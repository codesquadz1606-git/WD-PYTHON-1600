```python
import copy

info=["Rishabh","Chandigarh",34,["React","Node","Express"]]
# newData=info # this copies the element as well as the reference.
# newData=info.copy() # Shallow Copy : This copies the first level of data without any previous references.
newData=copy.deepcopy(info) # Deep Copy : This copies all the levels without any previous  references.


newData[0]="Shivangi"
newData[2]=25

newData[3][0]="Python"
newData[3][1]="Excel"
newData[3][2]="PowerBi"

print("Old",info)
print("New",newData)
```
