What is dictonary?
It is a key value pair storage , where we store elements in key:value.
{key:value,......}

```python
detail={
    "name":"Subhash",
    "city":"Noida",
    "blood_group":'O+'
}

print(detail)
print(type(detail))
```

```python
# Accessing the key from dictonary.

# Method 1 : using [keyname]
print(detail["name"]) # error

# Method 2 : using get(keyname)
print(detail.get("age")) # none
```

```python
# How we can add or update any key from dictonary?
detail["city"]="Delhi" # agar koi key mil gaya toh update ho jaygaa.
detail["age"]=30 # if key nhii milla toh update ho jaygaa.
print(detail)
```

```python
# I have to create 5 users in a list with their following details : name, age , city
data=[]
for i in range(5):
    info={}
    name=input("Enter your name")
    age=int(input("Enter your age"))
    city=input("Enter your city")

    info["name"]=name
    info["age"]=age
    info["city"]=city

    data.append(info)

print(data)

```

```python
for i in range(len(data)):
    print(data[i])
```
